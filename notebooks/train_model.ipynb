```python
import yfinance as yf
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib

print("📥 Downloading TCS Data...")
df = yf.download("TCS.NS", start="2020-01-01")

# Use only Close column
df = df[['Close']].copy()

# Create Target column (next day price)
df['Target'] = df['Close'].shift(-1)
df = df.dropna()

# Split into X and y
X = df[['Close']]
y = df['Target']

# Train model
model = LinearRegression()
model.fit(X, y)

# Save model
joblib.dump(model, "../models/tcs_sklearn_model.pkl")

print("✅ Model trained and saved as: models/tcs_sklearn_model.pkl")
print("🎉 Training complete! You can now run the Streamlit app.")
```
